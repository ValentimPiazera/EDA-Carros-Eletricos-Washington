## Deixando as células da coluna de fornecedor de energia mais intuitivas

 Decidi remover certas nominações e símbolos presentes nas células, para deixa-las mais curtas e intuitivas. Olhando para o caso de *Bonneville* vemos um caso interessante, pois como citado acima é uma companhia estatal, logo os *PUDs*(*Public Utility District*), que são postos estatais que recebem energia da *Bonneville*.

```python
# Limpando os nomes dos fornecedores de energia, removendo coisas com inc, WA, símbolos e nomes desnecessáriamente longos


#Retirando WA
df["Electric Utility"] = df["Electric Utility"].str.replace(" - (WA)", "", regex=False)

# Retirando - INC
df["Electric Utility"] = df["Electric Utility"].str.replace(" - INC", "", regex=False)

# Retirando INC
df["Electric Utility"] = df["Electric Utility"].str.replace("INC", "", regex=False)

# Retirando barras
df["Electric Utility"] = df["Electric Utility"].str.split("|").str[0]

# padronizando as células
df["Electric Utility"] = df["Electric Utility"].str.strip().str.title()

df["Electric Utility"] = df["Electric Utility"].str.strip()

mapeamento_fornecedores = {"NO KNOWN ELECTRIC UTILITY SERVICE": "Unknown","OKANOGAN COUNTY ELEC COOP,": 
                            "OKANOGAN COUNTY ELEC COOP",
                            "BONNEVILLE POWER ADMINISTRATION": "BPA (Local)"}

df["Electric Utility"] = df["Electric Utility"].replace(mapeamento_fornecedores)

df["Electric Utility"].unique()
```

## Verificando outliers

 Para verificar a existência de *outliers* eu decidi identificar por meio do método de quartis, que se trata de encontrar os valores mais baixos(25% mais baixos) e mais altos(75% mais altos) e compará-los em relação a média das informações. 

```python
# Função para identificar outliers pelo método de quartis

def outliers_quartile(dataframe, coluna):
    """Função para identificar outliers com o método de boxplot/quartiles
    Troubleshooting: 1- Pandas não foi importado
    2- A variável coluna não foi escrito como string"""

    Q1 = dataframe[coluna].quantile(.25)
    Q3 = dataframe[coluna].quantile(.75)
    IQR = Q3 - Q1
    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR
    outliers = dataframe[(dataframe[coluna] < limite_inferior) | 
                         (dataframe[coluna] > limite_superior)]
    
    return outliers
```

## *Outliers* mas nem tanto...

 Como dito anteriormente, ao pesquisar por *outliers* na coluna de idade do veículo, se encontra valores como o do Ford Ranger, que foi lançado em 1999, porém não removi ele e outras *outliers*, visto que não é algo mentiroso, apenas algo comum em um *dataset* que abrange tantos anos.

```python
# Verificando outliers nas colunas de idade, autonomia e ano de fabricação


outliers_quartile(df, "Vehicle Age")
```

```python
# Devido a muitos valores NaNs, foi retornado um Dataframe vazio
# Por segurança irei verificar de outra forma abaixo

outliers_quartile(df, "Electric Range")
```
